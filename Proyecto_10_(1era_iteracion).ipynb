# Parte 1 Priorizar hipótesis

Es momento de priorizar las hipótesis para averiguar cual de todas es la más viable y por lo tanto la que desarrollaremos. Voy a utilizar la metodología RICE para determinar cual es la hipótesis más viable.

```python
# Creo las columnas ICE y RICE para cada una de las hipótesis.

hypotheses['ICE'] = ((hypotheses['impact'] * hypotheses['confidence']
)/hypotheses['effort']).round(2)

hypotheses['RICE'] = (
    hypotheses['reach'] * hypotheses['impact'] * hypotheses['confidence']
)/ hypotheses['effort']

hypotheses
```

```python
# Hipótesis con el ICE más alto

hypotheses[['hypothesis', 'reach', 'impact', 'confidence', 'effort', 'ICE']].sort_values(by='ICE', ascending=False)
```

```python
# Hipótesis por el valor RICE más alto.

hypotheses[['hypothesis', 'reach', 'impact', 'confidence', 'effort', 'RICE']].sort_values(by='RICE', ascending=False)
```

**Conclusion de la priorización de hipótesis**

La hipótesis número 8 parecía ser la mejor con el enfoque ICE, pero el hecho que su alcance (reach) sea tan bajo (de 1) hace que se desplome y caiga muchas posiciones en el enfoque RICE, quiere decir que era una hipótesis bien elaborada, con un grado alto de confianza e impacto, pero iba a abarar a muy pocas personas.

La hipótesis número 0 es una buena prospecta, tiene valores bastante altos en ambas escalas (ICE y RICE), excelente impacto y alto grado de confianza, sin embargo su alcance también es bajo y su esfuerzo no es tan bajo, además hay una hipótesis que le supera en la escala RICE.

La hipótesis número 7, es la que tiene el valor RICE más alto por sobre todas las demás con 112 puntos, tiene el doble de puntaje que la hipótesis número 0. Y se encuentra entre las 3 hipótesis con el valor ICE más alto, es por ello que es la hipótesis que he decidido investigar.

## Ingreso acumulado por grupo

Observemos como fueron los ingresos para cada grupo.

```python
# Ingresos por grupo
revenue = orders.groupby('group')['revenue'].mean().reset_index()

fig = px.bar(
revenue, x='group', y='revenue',
width=800, height=600, color="group", title="El grupo B produjo mayores ingresos totales"
)

fig.show()
```

```python
fig = px.histogram(
    orders, 
    x='revenue', 
    color='group',  # Diferenciar por grupo A/B
    barmode='overlay',  # Superponer las distribuciones
    marginal='box',  # Agregar un boxplot en la parte superior (opcional)
    width=800, 
    height=600, 
    title="Distribución de ingresos por grupo"
)

# Mostrar el gráfico
fig.show()
```

El grupo B produjo mayores ingresos, sin embargo habrá que analizar más minuciosamente, dependiendo del número de personas en cada grupo y después de eliminar valores de anómalos, podríamos tener difenrentes resultados al observar el consumo promedio de las personas de cada grupo. 

```python
# Cambio las columnas date al tipo datetime 
orders['date'] = pd.to_datetime(orders['date'])
visits['date'] = pd.to_datetime(visits['date'])
```

```python
# Métrica acumulada de transacciones, visitantes y ganancia por día para cada grupo.

datesGroups = orders[['date','group']].drop_duplicates()
ordersAggregated = datesGroups.apply(lambda x: orders[np.logical_and(orders['date'] <= x['date'], orders['group'] == x['group'])].agg({'date' : 'max', 'group' : 'max', 'transactionId' : pd.Series.nunique, 'visitorId' : pd.Series.nunique, 'revenue' : 'sum'}), axis=1).sort_values(by=['date','group'])

ordersAggregated.head()
```

```python
# Métrica acumulada de visitas por día para cada grupo.

visitorsAggregated = datesGroups.apply(lambda x: visits[np.logical_and(visits['date'] <= x['date'], visits['group'] == x['group'])].agg({'date' : 'max', 'group' : 'max', 'visits' : 'sum'}), axis=1).sort_values(by=['date','group'])
visitorsAggregated
```

```python
#  Fusiono las dos tablas en una y coloco a sus columnas nombres descriptivos.

cumulativeData = ordersAggregated.merge(visitorsAggregated, left_on=['date', 'group'], right_on=['date', 'group'])
cumulativeData.columns = ['date', 'group', 'orders', 'buyers', 'revenue', 'visitors']
```

```python
# Gráfica de líneas de la ganancia acumulada.

fig = px.line(
    cumulativeData, 
    x='date', 
    y='revenue', 
    color='group',  # Diferenciar los grupos con colores
    title='El grupo B tiene una mayor cantidad de ingresos (A/B Test)',
    line_shape='linear',  # Asegura líneas rectas
    markers=True  # Mostrar puntos en las líneas
)

# Personalizar la leyenda (arriba a la izquierda)
fig.update_layout(
    legend_title="Grupo",
    legend=dict(
        x=0.02,   # Posición horizontal (izquierda)
        y=0.98,   # Posición vertical (arriba)
        xanchor="left",
        yanchor="top"
    ),
    xaxis_title="Fecha", 
    yaxis_title="Ingresos Acumulados ($)",
)

fig.show()

```

**Ganancia acumulada de la prueba**

En el gráfico podemos observar como la ganancia acumulada es mayor en el grupo B, sin embargo valdría la pena averiguar qué pasó el 19 de agosto, parece que hubo una cantidad de compras atípica en el grupo B, si descubrimos que las compras de esa fecha pertenecen a valores anómalos, quizá la brecha entre el grupo A y grupo B podría no ser tan grande ya que los gráficos son bastante similares, parece que el día 19 de agosto hizo que subiera mucho la gráfica pero podría tratarse de un valor atípico que podríamos rechazar para tener una aproximación más precisa.

## Diferencia relativa

```python
# Separo el df por grupos 

groupA = cumulativeData[cumulativeData['group'] == 'A']
groupB = cumulativeData[cumulativeData['group'] == 'B']

# uno el df por fecha

mergedCumulativeRevenue = groupA.merge(
    groupB, 
    on='date', 
    how='left', 
    suffixes=['A', 'B']
)

# Calculo la diferencia relativa en el tamaño promedio de compra
mergedCumulativeRevenue['relative_difference'] = (
    (mergedCumulativeRevenue['revenueB'] / mergedCumulativeRevenue['ordersB']) /
    (mergedCumulativeRevenue['revenueA'] / mergedCumulativeRevenue['ordersA']) - 1
)

mergedCumulativeRevenue.head()
```

```python
# Creo una gráfica para mostrarlo.

# Crear el gráfico de líneas con Plotly Express
fig = px.line(
    mergedCumulativeRevenue, 
    x='date', 
    y='relative_difference', 
    title='El grupo B tiene un mejor rendimiento en términos de ingresos promedio por pedido.',
    labels={'date': 'Fecha', 'relative_difference': 'Diferencia Relativa'},
    line_shape='linear',  # Líneas rectas
    markers=True  # Agregar puntos en la línea
)

# Agregar una línea horizontal de referencia en y=0
fig.add_hline(y=0, line_dash="dash", line_color="black", annotation_text="Referencia")

# Personalizar el diseño del gráfico
fig.update_layout(
    xaxis_title="Fecha",
    yaxis_title="Diferencia Relativa",

)

# Mostrar el gráfico
fig.show()
```

Este gráfico nos muestra que el grupo B tiene un mejor rendimiento, aunque me sigue causando ruido el 19 de agosto, que podría estar modificando nuestros resultados

## significancia estadística de la diferencia en la conversión entre los grupos (datos en bruto)

```python
# agrupo los usuarios que hicieron ordenes en A y B

ordersByUsersA = orders[orders['group']=='A'].groupby('visitorId', as_index=False).agg({'transactionId' : pd.Series.nunique})
ordersByUsersA.columns = ['userId', 'orders']

ordersByUsersB = orders[orders['group']=='B'].groupby('visitorId', as_index=False).agg({'transactionId' : pd.Series.nunique})
ordersByUsersB.columns = ['userId', 'orders']
```

```python
# Agrego a cada grupo la cantidad de personas que entraron pero no realizaron ordenes (orders = 0)

sampleA = pd.concat([ordersByUsersA['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='A']['visits'].sum() - len(ordersByUsersA['orders'])), name='orders')],axis=0)

sampleB = pd.concat([ordersByUsersB['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='B']['visits'].sum() - len(ordersByUsersB['orders'])), name='orders')],axis=0)
```

```python
# Calculo si existe diferencia estadística entre los grupos A y B.

p_value = round(stats.mannwhitneyu(sampleA, sampleB)[1], 3)

print('P-value:', p_value)

if p_value < 0.05:
    print("Existe diferencia significativa entre los grupos (P-value < 0.05)")
else:
    print("No hay diferencia significativa entre los grupos (P-value >= 0.05)")


# Calculo la diferencia relativa entre las medias de sampleB y sampleA
relative_difference = round((sampleB.mean() / sampleA.mean() - 1), 3)

print('\nComparación relativa:', relative_difference)

if relative_difference > 0:
    print(f"La media de B es mayor que A en un {relative_difference * 100}%")
else:
    print(f"La media de B es menor o igual que A en un {relative_difference * 100}%")
```

Con los datos sin procesar tenemos una diferencia estadística que nos indica que los grupos son diferentes, y que la media del grupo B es mayor a la del A en un 13.8%

## Significancia estadística de la diferencia en el tamaño promedio de pedido entre los grupos (datos en bruto)

```python
# Comparo el tamaño promedio de los pedidos para ambos grupos.

p_value = round(stats.mannwhitneyu(
    orders[orders['group'] == 'A']['revenue'], 
    orders[orders['group'] == 'B']['revenue']
)[1], 3)

# Calculo la diferencia relativa de las medias de ingresos.

relative_difference = round((orders[orders['group'] == 'B']['revenue'].mean() / \
                      orders[orders['group'] == 'A']['revenue'].mean() - 1), 3)


print(f"P-value de la prueba de Mann-Whitney U: {p_value:.3f}")

if p_value < 0.05:
    print("Existe una diferencia estadísticamente significativa entre los grupos (P-value < 0.05)")
else:
    print("No se encontró una diferencia estadísticamente significativa (P-value >= 0.05)")

print(f"\nDiferencia relativa de las medias: {relative_difference:.3f}")

if relative_difference > 0:
    print(f"El grupo B tiene un ingreso promedio {relative_difference:.1%} mayor que el grupo A.")
else:
    print(f"El grupo A tiene un ingreso promedio {-relative_difference:.1%} mayor que el grupo B.")
```

Respecto a los tamaños promedios en los pedidos realizados en los grupos, la prueba (Mann-Whitney U) no encontró evidencia suficiente para decir que los grupos son estadísticamente diferentes, sin embargo la diferencia relativa de las medias indica que la media del grupo B es más del 25% mayor que la media del grupo A.

```python
# Encuentro a los usuarios con más de 2 órdenes.
usersWithManyOrders = pd.concat([ordersByUsersA[ordersByUsersA['orders'] > 2]['userId'], ordersByUsersB[ordersByUsersB['orders'] > 2]['userId']], axis = 0)

# Encuentro a los usuarios con compras de más de $436.
usersWithExpensiveOrders = orders[orders['revenue'] > 436]['visitorId']

# fusiono esas dos colecciones de ID's, y elimino duplicados.
abnormalUsers = pd.concat([usersWithManyOrders, usersWithExpensiveOrders], axis = 0).drop_duplicates().sort_values()

print(abnormalUsers.head(5))
print(abnormalUsers.shape)
```

Tenemos 73 usuarios anómalos, vamos a eliminarlos de nuestros datos, para tener los datos límpios de esos datos extremos.

```python
# Filtro los df quitando los usuarios con comportamientos anómalos.

sampleAFiltered = pd.concat([ordersByUsersA[np.logical_not(ordersByUsersA['userId'].isin(abnormalUsers))]['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='A']['visits'].sum() - len(ordersByUsersA['orders'])),name='orders')],axis=0)
sampleBFiltered = pd.concat([ordersByUsersB[np.logical_not(ordersByUsersB['userId'].isin(abnormalUsers))]['orders'],pd.Series(0, index=np.arange(visits[visits['group']=='B']['visits'].sum() - len(ordersByUsersB['orders'])),name='orders')],axis=0)
```

```python
# Calculo si existe diferencia estadística entre los grupos A y B.

p_value = round(stats.mannwhitneyu(sampleAFiltered, sampleBFiltered)[1], 3)

print('P-value:', p_value)

if p_value < 0.05:
    print("Existe diferencia significativa entre los grupos (P-value < 0.05)")
else:
    print("No hay diferencia significativa entre los grupos (P-value >= 0.05)")


# Calculo la diferencia relativa entre las medias de sampleB y sampleA
relative_difference = round((sampleBFiltered.mean() / sampleAFiltered.mean() - 1), 3)

print('\nComparación relativa:', relative_difference)

if relative_difference > 0:
    print(f"La media de B es mayor que A en un {relative_difference * 100}%")
else:
    print(f"La media de B es menor o igual que A en un {relative_difference * 100}%")
```
